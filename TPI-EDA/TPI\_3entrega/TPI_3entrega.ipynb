# 📁 Dataset seleccionado– Sleep Health and Lifestyle

## 📝 Descripción del Dataset

El conjunto de datos "Sleep Health & Daily Performance Dataset" \(publicado por Mohan Krishna Thalla en Kaggle; disponible en \(https://www\.kaggle\.com/datasets/mohankrishnathalla/sleep\-health\-and\-daily\-performance\-dataset\)\) reúne 100\.000 registros diarios y 32 variables orientadas al estudio de la relación entre hábitos de estilo de vida, factores laborales, aspectos psicológicos e indicadores del descanso\.  Las 32 variables se estructuran en cinco dimensiones documentadas por el autor:   

Sociodemográfica y laboral: person\_id \(identificador\), age \(edad en años, 18 a 69\), gender \(género\), country \(15 países\), occupation \(12 ocupaciones\), work\_hours\_that\_day \(horas trabajadas en el día\) y shift\_work \(trabajo por turnos: 0 = No, 1 = Sí\)\.   

Arquitectura e indicadores del sueño: sleep\_duration\_hrs \(horas dormidas\), sleep\_quality\_score \(puntaje de calidad percibida de 1 a 10\), rem\_percentage y deep\_sleep\_percentage \(porcentaje de sueño REM y profundo\), sleep\_latency\_mins \(minutos hasta conciliar el sueño\), wake\_episodes\_per\_night \(despertares nocturnos\), nap\_duration\_mins \(minutos de siesta\), weekend\_sleep\_diff\_hrs \(diferencia de horas en fin de semana\), chronotype \(cronotipo: matutino, vespertino o neutro\) y sleep\_aid\_used \(uso de ayuda para dormir: 0/1\)\.   

Hábitos y estilo de vida: caffeine\_mg\_before\_bed \(mg de cafeína antes de dormir\), alcohol\_units\_before\_bed \(unidades de alcohol\), screen\_time\_before\_bed\_mins \(minutos de pantalla previos a dormir\), exercise\_day \(realizó actividad física: 0 = No, 1 = Sí\) y steps\_that\_day \(pasos diarios\)\.   

Indicadores biométricos, psicológicos y contextuales: bmi \(índice de masa corporal\), heart\_rate\_resting\_bpm \(frecuencia cardíaca en reposo\), stress\_score \(nivel de estrés de 1 a 10\), mental\_health\_condition \(condición de salud mental reportada\), room\_temperature\_celsius \(temperatura ambiental\), season \(estación del año\) y day\_type \(día laboral o fin de semana\)\.   

Variables de resultado \(Targets potenciales\): felt\_rested \(sensación de despertar descansado: 0/1\), cognitive\_performance\_score \(puntaje de rendimiento cognitivo de 0 a 100\) y sleep\_disorder\_risk \(nivel de riesgo de trastorno del sueño: Healthy, Mild, Moderate, Severe\)\.   

Aclaración metodológica y limitaciones de los datos: Se trata de un conjunto de datos sintético cuyas distribuciones y correlaciones fueron generadas de forma programática a partir de parámetros de encuestas poblacionales \(CDC NHIS, Sleep Foundation\)\. En consecuencia, no constituye una muestra probabilística de una población real: resulta adecuado para prácticas de análisis exploratorio y modelado de aprendizaje automático, pero sus resultados representan asociaciones internas de la simulación y no pueden trasladarse directamente a una población real ni interpretarse como evidencia clínica\.

5\. Análisis de Outliers por Rango Intercuartílico \(IQR\)

```
print("=== COMPROBACIÓN DE VALORES ATÍPICOS (OUTLIERS MEDIANTE IQR) ===")

cols_outliers = [
    'sleep_duration_hrs',
    'caffeine_mg_before_bed',
    'screen_time_before_bed_mins',
    'stress_score',
]

for col in cols_outliers:
    if col not in df_salud.columns:
        print(f"No se encontró la columna '{col}' en df_salud.")
        continue

    # Se excluyen valores nulos para calcular los cuartiles y el porcentaje.
    valores = df_salud[col].dropna()
    if valores.empty:
        print(f"La columna '{col}' no contiene valores válidos.")
        continue

    Q1 = valores.quantile(0.25)
    Q3 = valores.quantile(0.75)
    IQR = Q3 - Q1
    lim_inf = Q1 - 1.5 * IQR
    lim_sup = Q3 + 1.5 * IQR

    outliers = valores[(valores < lim_inf) | (valores > lim_sup)]
    pct = (len(outliers) / len(valores)) * 100

    print(f"\nVariable '{col}':")
    print(f" • Q1 (25%): {Q1:.2f} | Q3 (75%): {Q3:.2f} | IQR: {IQR:.2f}")
    print(f" • Límites teóricos: [{lim_inf:.2f}, {lim_sup:.2f}]")
    print(f" • Outliers detectados: {len(outliers)} registros ({pct:.2f}% de los valores válidos)")
```

Se evaluó la matriz completa de datos mediante la aplicación de las funciones \`isnull\(\)\.sum\(\)\` y \`duplicated\(\)\.sum\(\)\` confirmando la integridad técnica del dataset en sus 100\.000 observaciones\. La ausencia de datos faltantes \(0 nulos\) y de filas repetidas \(0 duplicados\) garantiza que no sea necesario aplicar técnicas de imputación ni eliminación de registros que puedan sesgar la distribución original\.

Esto descarta la presencia de valores imposibles, errores de escala o fallas de carga\.

Se inspeccionaron la totalidad de las variables categóricas de texto \(\`occupation\`, \`country\`, \`gender\`, \`chronotype\`, \`sleep\_disorder\_risk\`, \`mental\_health\_condition\`, \`season\` y \`day\_type\`\), confirmando que todas las etiquetas se encuentran estandarizadas, sin inconsistencias de formato, espacios en blanco ni errores de tipografía\.

## Análisis Bivariado

# Nivel de Estrés vs\. Calidad del Sueño

Se explora la relación bivariada entre el puntaje de estrés percibido \(\`stress\_score\`\) y la calidad del descanso \(\`sleep\_quality\_score\`\) mediante un gráfico de dispersión con ajuste de regresión lineal\.

```
plt.figure(figsize=(10, 5))
sns.regplot(data=df_salud.sample(2000, random_state=42), x='stress_score', y='sleep_quality_score', scatter_kws={'alpha': 0.3, 'color': 'steelblue'}, line_kws={'color': 'firebrick', 'linewidth': 2})
plt.title('Relación Bivariada: Nivel de Estrés vs. Calidad del Sueño', fontsize=12, fontweight='bold') 
plt.xlabel('Puntaje de Estrés (stress_score)', fontsize=10) 
plt.ylabel('Calidad del Sueño (sleep_quality_score)', fontsize=10) 
plt.grid(True, linestyle='--', alpha=0.5)
r = df_salud['stress_score'].corr(df_salud['sleep_quality_score']) 
print(f"Coeficiente de Correlación de Pearson (r): {r:.2f}")
plt.show()
```

Se observa una tendencia lineal descendente: a medida que los valores de estrés son más altos, el puntaje de calidad del sueño tiende a ser menor\. El coeficiente de correlación de Pearson \(r = \-0\.64\) indica una asociación negativa moderada\-fuerte entre ambas variables dentro del dataset\.

# Procesamiento de datos \(Encoding, Split Y Escalado\)

### Definición y Selección de la Variable Objetivo \(\`sleep\_disorder\_risk\`\)

Previo a la fase de modelado predictivo, se define como variable objetivo \(target\) a sleep\_disorder\_risk\. Esta variable categórica clasifica el nivel de riesgo de trastorno del sueño en cuatro niveles: Healthy\(Sin riesgo/Sano\), Mild \(Riesgo Leve\), Moderate \(Riesgo Moderado\) y Severe \(Riesgo Severo\)\.

Justificación de la elección:

Impacto Preventivo y Clínico: A diferencia de indicadores subjetivos o generales del dataset \(como la percepción de descanso \`felt\_rested\` o el rendimiento cognitivo \`cognitive\_performance\_score\`\), \`sleep\_disorder\_risk\` constituye la etiqueta diagnóstica por excelencia para habilitar sistemas de alerta temprana y medicina preventiva\.

Reflejo de la Prevalencia Real: Aunque la variable presenta un desbalance de clases \(donde las categorías de bajo riesgo representan el 87\.6% de la muestra y los casos moderados/severos concentran el 12\.4%\), esta distribución no es un defecto de los datos, sino un reflejo fiel de la realidad epidemiológica en la población general\. Enfrentar este desbalance permite construir un modelo preparado para escenarios de aplicación real\.

### Metodología de Preprocesamiento de Datos

Para transformar los datos crudos en tensores numéricos optimizados para los algoritmos de Machine Learning \(Gradient Boosting, Random Forest, Regresión Logística y KNN\), se ejecutan tres transformaciones secuenciales:

1\. Codificación de Variables Categóricas \(One\-Hot Encoding\): Se transforman los atributos cualitativos \(\`gender\`, \`occupation\`, \`country\`, \`chronotype\` y \`season\`\) en variables binarias \(dummy variables\) mediante \`pd\.get\_dummies\(\.\.\., drop\_first=True\)\` para evitar problemas de multicolinealidad\.
2\. División Estratificada del Dataset \(Stratified Train/Test Split\): Se divide la muestra en un 80% para entrenamiento y un 20% para evaluación independiente \(prueba\)\. Se utiliza muestreo estratificado \(\`stratify=y\`\) para garantizar que la proporción del 12\.4% de clases de alto riesgo se conserve exactamente igual en ambos subconjuntos\.
3\. Escalado de Atributos Numéricos \(StandardScaler\): Se estandarizan las características continuas \(como edad, horas de sueño, nivel de estrés y frecuencia cardíaca\) restando la media y dividiendo por la desviación estándar\. El escalador se ajusta exclusivamente sobre el conjunto de entrenamiento \(\`fit\_transform\`\) y se aplica posteriormente sobre el conjunto de prueba \(\`transform\`\), previniendo cualquier tipo de fuga de información \(data leakage\)\.

```
TARGET = 'sleep_disorder_risk'
features_num = [ 
    'age', 
    'sleep_duration_hrs', 
    'sleep_quality_score', 
    'stress_score', 
    'heart_rate_resting_bpm', 
    'caffeine_mg_before_bed', 
    'screen_time_before_bed_mins', 
    'alcohol_units_before_bed', 
    'exercise_day', 
    'steps_that_day', 
    'work_hours_that_day',
]
features_cat = ['gender', 'occupation', 'country', 'chronotype', 'season']
X = df_salud[features_num + features_cat].copy() 
y = df_salud[TARGET].copy()
```

1\. One\-Hot Encoding

```
X_encoded = pd.get_dummies(X, columns=features_cat, drop_first=True)
```

2\. División Stratified Train \(80%\) / Test \(20%\)

```
X_train, X_test, y_train, y_test = train_test_split( X_encoded, y, test_size=0.20, random_state=42, stratify=y )
```

3\. Escalado numérico con StandardScaler \(Ajustado solo en Train\)

```
scaler = StandardScaler() 
X_train_scaled = X_train.copy() 
X_test_scaled = X_test.copy()
X_train_scaled[features_num] = scaler.fit_transform(X_train[features_num])
X_test_scaled[features_num] = scaler.transform(X_test[features_num])
```

```
print('✅ Preprocesamiento completado:') 
print(f'• Muestra de Entrenamiento (Train): {X_train.shape[0]:,} registros') 
print(f'• Muestra de Evaluación (Test): {X_test.shape[0]:,} registros') 
print(f'• Atributos procesados: {X_train.shape[1]} columnas')
```

## Entrenamiento y comparación de 4 modelos

```
modelos = { 
    'Regresión Logística': LogisticRegression( max_iter=1000, random_state=42, multi_class='multinomial' ), 
    'K-Nearest Neighbors (KNN)': KNeighborsClassifier(n_neighbors=5), 
    'Random Forest': RandomForestClassifier(n_estimators=100, random_state=42), 
    'Gradient Boosting': GradientBoostingClassifier( n_estimators=100, random_state=42 ), 
}
resultados = []
for nombre, modelo in modelos.items(): 
    modelo.fit(X_train_scaled, y_train) 
    y_pred = modelo.predict(X_test_scaled)

    resultados.append({ 
    'Modelo': nombre, 'Accuracy': accuracy_score(y_test, y_pred), 
    'Precision': precision_score(y_test, y_pred, average='weighted'), 
    'Recall': recall_score(y_test, y_pred, average='weighted'), 
    'F1-Score': f1_score(y_test, y_pred, average='weighted'), }) 
df_resultados = ( pd.DataFrame(resultados) .sort_values(by='F1-Score', ascending=False) .reset_index(drop=True) ) 
print('=== 📊 TABLA COMPARATIVA DE MÉTRICAS EN TEST ===') 
display(df_resultados.style.highlight_max(color='#d4edda', axis=0))
```

### Matriz de Confusión y Reporte de Clasificación \(Random Forest\)

```
rf_modelo = RandomForestClassifier(n_estimators=100, random_state=42)
rf_modelo.fit(X_train_scaled, y_train)
y_pred_rf = rf_modelo.predict(X_test_scaled)

clases = sorted(pd.Series(y_test).dropna().unique())
cm = confusion_matrix(y_test, y_pred_rf, labels=clases)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt=',d',
    cmap='Blues',
    xticklabels=clases,
    yticklabels=clases,
    annot_kws={'size': 11, 'weight': 'bold'},
)
plt.title(
    'Matriz de Confusión - Random Forest ',
    weight='bold',
    pad=12,
)
plt.xlabel('Predicción del Modelo', weight='bold')
plt.ylabel('Diagnóstico Real', weight='bold')
plt.tight_layout()
plt.show()

print('=== 📋 REPORTE DE CLASIFICACIÓN POR CLASE ===')
print(classification_report(y_test, y_pred_rf, labels=clases, zero_division=0))
```

El desarrollo del sistema predictivo de riesgo de trastornos del sueño \(\`sleep\_disorder\_risk\`\) permitió evaluar y comparar cuatro algoritmos de aprendizaje automático supervisado \(Gradient Boosting, Random Forest, Regresión Logística y K\-Nearest Neighbors\)\.

1\. Selección del Modelo y Factores Clave

Gradient Boosting Classifier \(seguido de cerca por Random Forest\) se consolidó como el modelo con mejor desempeño global, alcanzando una Accuracy del 73\.95% y un F1\-Score ponderado de 0\.73 sobre la muestra de prueba independiente de 20\.000 observaciones\. El análisis de importancia de características \(Feature Importance\) en Gradient Boosting reveló que más del 94% del poder predictivo del modelo se concentra en tres variables centrales: calidad percibida del sueño \(74\.0%\), duración del descanso \(16\.7%\) y nivel de estrés \(4\.2%\), seguidas de forma marginal por el consumo de alcohol antes de dormir \(1\.4%\), la edad \(1\.2%\) y el consumo de cafeína \(1\.2%\)\. Por su parte, en Random Forest la importancia se distribuye de forma más repartida, liderada igualmente por la calidad del sueño \(22\.9%\), duración del sueño \(15\.1%\), nivel de estrés \(10\.6%\) y horas de trabajo \(6\.2%\)\.

2\. Análisis  y Salvedades sobre el Desempeño por Clase

Si bien el desempeño global del modelo es satisfactorio, la evaluación desglosada por categorías de riesgo expone una limitación crítica vinculada al desbalance de clases \(class imbalance\):

\*Clases Mayoritarias \(Healthy y Mild \): Concentran el 87\.6% de la muestra total\. El modelo demuestra una alta efectividad para identificar individuos sanos \(F1\-Score = 0\.86, Recall = 89%\) y con riesgo leve \(F1\-Score = 0\.66, Recall = 68%\)\.

\*Clases Minoritarias \(Moderate y Severe \): Representan únicamente el 12\.4%de los datos\. Esto genera que la Accuracy global \(74%\) se encuentre inflada por el peso de la clase mayoritaria\. Para la categoría Moderate, el modelo alcanza un F1\-Score bajo de 0\.26 \(Recall de 0\.19 y Precisión de 0\.40\), mientras que para la categoría Severe alcanza un F1\-Score de 0\.52 \(Recall de 0\.45\)\. Esto significa que el algoritmo tiende a subestimar el riesgo en pacientes con síntomas moderados, confundiéndolos con categorías de menor severidad\.

El desbalance observado en la variable objetivo \(donde las categorías de bajo riesgo representan el 87\.6% de la muestra y las categorías \`Moderate\` y \`Severe\` concentran únicamente el 12\.4%\) no constituye un defecto del dataset, sino un reflejo fiel de la distribución epidemiológica real\.  En el ámbito de la salud clínica, la prevalencia de patologías moderadas a severas es por definición minoritaria\. Seleccionar esta variable garantiza que el sistema enfrente y resuelva la complejidad real del dominio, en lugar de simplificar artificialmente el problema con variables sintéticas o menos relevantes\.

3\. Propuestas de Mejora y Trabajo Futuro

Para subsanar el impacto del desbalance de clases y optimizar la sensibilidad clínica del sistema en la detección de casos graves, se proponen las siguientes líneas de acción técnica para futuras iteraciones:

\-Ponderación de Clases y Muestras \(class\_weight / sample\_weight\): Configurar la función de pérdida para penalizar con mayor severidad los errores en los grupos Moderate y Severe, utilizando class\_weight='balanced' en Random Forest y Regresión Logística, o aplicando vectores de pesos ponderados \(compute\_sample\_weight\('balanced', y\_train\)\) durante el entrenamiento de Gradient Boosting\.

\-Técnicas de Re sampleo \(SMOTE\): Aplicar técnicas de sobre muestreo sintético sobre la muestra de entrenamiento para nivelar la cantidad de registros en las categorías de mayor riesgo previo al ajuste de los modelos\.

\-Ajuste del Umbral de Decisión \(Threshold Tuning\): Modificar los umbrales de probabilidad requeridos para activar las alertas preventivas, priorizando la métrica de Recall \(Sensibilidad\) por sobre la precisión para minimizar la tasa de Falsos Negativos en diagnósticos médicos\.
