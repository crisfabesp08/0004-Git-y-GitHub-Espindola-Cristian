```
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

datos = np.array([10, 20, 30, 40, 50])
print(datos)
```

```
data = {
    "estudiante": ["Jose", "Cris", "Carla", "Ana", "Maria"],
    "nota": [2, 4, 8, 10, 7]
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)

promedio = df["nota"].mean()
print(f"Promedio de notas: {promedio}")

nota_max = df["nota"].max()
print(f"Nota máxima: {nota_max}")

nota_min = df["nota"].min()
print(f"Nota mínima: {nota_min}")
```

1\)pandas: porque permite leer archivos CSV y manipular datasets tabulares\.  
2\)seaborn: porque facilita la creación de gráficos estadísticos y distribuciones de datos\.  
3\)scikit\-learn: porque es una librería especializada en Machine Learning, con algoritmos de clasificación, regresión y clustering\.  
4\) numpy: porque ofrece operaciones matemáticas eficientes sobre arrays y matrices\.
